# `*`和`**`的用法
### 1. 定义函数时

```python
def wrapper(*args, **kwargs):
    # 能接收任意数量、任意类型参数
    pass
```

- `*args` --- 收集所有多余的 **位置参数**，打包成元组`tuple`
- `**kwargs` --- 收集所有多余的 **关键字参数**，打包成字典`dict`

### 2. 调用函数 / 拆包数据时

1. `*`解包列表、元组、可迭代对象

```python
# 将序列一个个拆开，变成独立位置参数
def add(x, y):
    return x + y

nums = [3, 5]
print(add(*nums))  # 等价 add(3,5) → 8
```

2. `**`解包字典

```python
# 把字典的键值对拆开，变成关键字参数
def info(name, age):
    print(name, age)

d = {"name": "V", "age": 19}
info(**d)  # 等价 info(name="V", age=19)
```

### 三、其他常见小用法
1. 变量拆分取值

```python
lst = [1, 2, 3, 4]
first, *rest = lst
print(first)  # 1
print(rest)   # [2, 3, 4]
```

2. 拼接多个列表

```python
a = [1, 2]
b = [3, 4]
c = [*a, *b]
print(c)  # [1,2,3,4]
```

3. 拼接字典（Python3.5+）

```python
d1 = {"a": 1}
d2 = {"b": 2}
d3 = {**d1, **d2}
print(d3) # {'a': 1, 'b': 2}
```
