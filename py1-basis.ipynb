|函数   |功能   |

|`chr()`|把编码转换为对应的字符|

==str是不变对象==
所以split和join都是返回新字符串

```python
print(ord('A'))
print(chr(66))
```

```python
# str的不可变
a = 'abc'
a.replace('a','A')
print(a)

b = a.replace('a','A') # replace只是返回一个新字符串，实际上a并没有改变
print(b)
```

```python
# split方法的用法示例
# 格式：
# str.split(sep=None, maxsplit=-1)  # sep：分隔符（可选）；maxsplit：最大分割次数（可选）(默认-1即全部分割)

# 1. 默认分割（空白分割）
s = "I love python"
res = s.split()
print(res)

# 2. 指定分割符分割
s2 = "苹果,香蕉,橘子"
print(s2.split(','))

# 3. 限制分割次数 maxsplit
s3 = "1-2-3-4-5"
# 只分割2次
print(s3.split('-', maxsplit=2))

# 4. 如果字符串开头 / 结尾就是分隔符，分割会出现空串：
s4 = ",a,b,c,"
print(s4.split(','))
```

```python
# join用法示例
lst = ['123', '456']
res = '.'.join(lst)
print(res)  # 123.456
```

## Python的编码
|方法   |功能   |
|:------:|:-------:|
|`encode()`|将以Unicode表示的str编码为指定的bytes|
|`decode()`|把bytes变为str|

```python
print('ABC'.encode('ascii'))
print('中文'.encode('utf-8'),'\n')

print(b'ABC'.decode('ascii'))
print(b'\xe4\xb8\xad\xe6\x96\x87'.decode('utf-8'))

# 如果bytes中只有一小部分无效的字节，可以传入errors='ignore'忽略错误的字节
b'\xe4\xb8\xad\xff'.decode('utf-8', errors='ignore')
```

## 格式化
|占位符   | 	替换内容  |
|:------:|:-------:|
|`%d`|整数|
|`%f`|浮点数|
|`%s`|字符串|
|`%x`|十六进制整数|

```python
>>> 'Hello, %s' % 'world'
'Hello, world'
>>> 'Hi, %s, you have $%d.' % ('Michael', 1000000)
'Hi, Michael, you have $1000000.'

>>> 'growth rate: %d %%' % 7  # 字符串里面的%是一个普通字符就需要转义，用%%来表示一个%
'growth rate: 7 %'
```

- 除此此外还有两种方法
|方法   |功能   |

```python
>>> 'Hello, {0}, 成绩提升了 {1:.1f}%'.format('小明', 17.125)
'Hello, 小明, 成绩提升了 17.1%'

r = 2.5
s = 3.14 * r ** 2
print(f'The area of a circle with radius {r} is {s:.2f}')
```

# 5.5 模式匹配
- 一下分别列举**简单匹配**、**复杂匹配**、**匹配列表**

```python
score = 'B'

match score:
    case 'A':
        print('score is A.')
    case 'B':
        print('score is B.')
    case 'C':
        print('score is C.')
    case _: # _表示匹配到其他任何情况
        print('score is ???.')
```

```python
age = 15

match age:
    case x if x < 10:    # 表示当age < 10成立时匹配，且赋值给变量x
        print(f'< 10 years old: {x}')
    case 10:             # 仅匹配单个值
        print('10 years old.') 
    case 11 | 12 | 13 | 14 | 15 | 16 | 17 | 18:   # 匹配多个值
        print('11~18 years old.')
    case 19:
        print('19 years old.')
    case _:
        print('not sure.')
```

```python
args = ['gcc', 'hello.c', 'world.c']
# args = ['clean']
# args = ['gcc']

match args:
    # 如果仅出现gcc，报错:
    case ['gcc']:
        print('gcc: missing source file(s).')
    # 出现gcc，且至少指定了一个文件:
    case ['gcc', file1, *files]:   # 表示列表第一个字符串是'gcc'，第二个字符串绑定到变量file1，后面的任意个字符串绑定到*files
        print('gcc compile: ' + file1 + ', ' + ', '.join(files))
    # 仅出现clean:
    case ['clean']:
        print('clean')
    case _:
        print('invalid command.')
```

# 5.7 快速查找 --- dict和set
|类型   |功能   |  对应方法|
|:------:|:-------:|:-------:|
|`dict`|key-value存储方式|①通过`in`判断key是否存在<br>②`get()`方法，如果key不存在，可以返回None，或者自己指定的value<br>③`pop(key)`方法----删除一个key|
|`set`|①是一组key的集合，但不存储value<br>②重复元素在set中自动被过滤<br>③可以看成数学意义上的无序和无重复元素的集合，可以做数学意义上的交集、并集等操作|语法就是`s = {1,2,3}`<br>add(key)方法---添加元素到set中<br>remove(key)方法----删除元素|

```python
# dict操作
d = {'Michael': 95, 'Bob': 75, 'Tracy': 85}
print(d['Bob'])

d['Adam'] = 67
print(d['Adam'])

print('Thomas' in d) # 1. 通过in判断key是否存在
print(d.get('Thomas')) # 2. 通过get()方法,不存在返回None
print(d.get('Thomas', -1)) # 返回自己指定的value

# pop()删除
d.pop('Bob')
print(d)
```

== dict和set的key必须是不可变对象 ==
- 在Python中，字符串、整数等都是不可变的，因此，可以放心地作为key。而list是可变的，就不能作为key：

```python
# dict的key值不可变案例
key = [1, 2, 3]
d[key] = 'a list'
```

```python
# set操作
s = {1, 1, 2, 2, 3, 3}
print(s)    # 重复元素在set中自动被过滤

# 添加
s.add(4)
print(s)

# 删除
s.remove(4)
print(s)
```

```python
# set 的交，并操作
s1 = {1, 2, 3}
s2 = {2, 3, 4}
print(s1 & s2)
print(s1 | s2)
```
