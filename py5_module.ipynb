# 模板
## 一、概念
- 一个.py文件就是一个**模块MOdule**,用**包**来按目录的形式组织模块

```
mycompany
├─ __init__.py
├─ abc.py
└─ xyz.py
```

- 此时abc.py的模块名就是`mycompany.abc`
- 请注意，每一个包目录下面都会有一个__init__.py的文件，这个文件是**必须存在**的，否则，Python就把这个目录当成普通目录，而不是一个包。
- `__init__.py`可以是空文件，也可以有Python代码，因为__init__.py本身就是一个模块，而它的模块名就是mycompany

==自己创建模块时要注意命名，**不能和Python自带的模块名称冲突**。例如，系统自带了sys模块，自己的模块就不可命名为sys.py，否则将无法导入系统自带的sys模块。==

## 二、模板标准写法

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

' a test module '

__author__ = 'Michael Liao'

import sys

def test():
    args = sys.argv
    if len(args)==1:
        print('Hello, world!')
    elif len(args)==2:
        print('Hello, %s!' % args[1])
    else:
        print('Too many arguments!')

if __name__=='__main__':
    test()
````

1. **头部注释行**
   - `#!/usr/bin/env python3`：Linux/macOS系统专属声明,告诉操作系统用系统环境里的 Python3 解释器运行该文件；Windows 系统无效
   - `# -*- coding: utf-8 -*-`：文件编码声明,指定源码使用 UTF-8 编码，支持中文，避免中文乱码。
2. **文档字符串**
   - `'a test module'`:任何模块代码的第一个字符串都被视为模块的文档注释,用于说明这个 Python 文件的用途，可通过工具查看模块说明。
3. **作者变量**
   - `__author__ = 'Michael Liao'`:Python 内置魔术变量，用来标注脚本作者信息。
4. **导入系统模块**
   - `import sys`:获取命令行运行时传入的参数列表。
5. **sys.argv**:命令行参数**字符串列表**
   - `sys.argv[0]`：永远是当前脚本的文件名（固定占用第一个位置）
   - `sys.argv[1]、sys.argv[2]...`：你手动输入的额外参数
6. **程序入口判断**
```
   if __name__=='__main__':
    test()
```
区分两种运行场景：<br>
    1. **直接运行**这个脚本文件（终端执行：python 文件名.py）
    此时__name__的值会被 Python 自动赋值为字符串'__main__'，条件成立，调用test()函数，代码执行生效<br>
    2. 把这个文件当作模块，被其他 Python 文件 import**导入调用**
    此时__name__等于本模块文件名，条件不成立，test()不会自动运行。<br>
    好处：既可以独立运行测试，又能当作工具模块被别的代码调用，互不干扰。

## 三、作用域
|类型|写法|外部能否直接访问|设计目的|
|:-----:|:--:|:------------:|:------------:|
|公有`public`|`data`|	✅完全随意读写|	对外开放，所有人可用|
|受保护`protected`|`_data`|✅ 语法可行，约定禁止外部调用|给子类用，外人别碰|
|私有`private`|`__data`|❌ 无法直接访问|绝密数据，只能类内部操作，外部通过方法访问修改|
|魔法属性|`__data__`|✅正常访问|Python 内置特殊方法 / 属性，不要自己定义这种名字|

````python
# 示例1
def _private_1(name):
    return 'Hello, %s' % name

def _private_2(name):
    return 'Hi, %s' % name

def greeting(name):
    if len(name) > 3:
        return _private_1(name)
    else:
        return _private_2(name)
````

````python
# 示例2
class User:
    def __init__(self):
        self.name = "小明"        # public 用户名（公开信息）
        self._level = 5          # protected 等级（子类可以继承修改）
        self.__password = "123456"# private 密码（隐私，绝不外露）
````
