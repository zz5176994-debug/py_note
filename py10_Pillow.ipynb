# Pillow 四大核心模块功能 + 可运行示例
## 1. Image（核心图像对象，负责图片创建、打开、保存、尺寸变换）
**职责**：管理整张图片本身，画布创建、读写文件、裁剪缩放、旋转、模式转换
|常用功能|示例代码|
|---|---|
|打开本地图片|`im = Image.open("test.jpg")`|
|创建空白画布(RGB彩色)|`im = Image.new("RGB", (200,100), (255,255,255))`|
|获取宽高尺寸|`w, h = im.size`|
|等比例缩小缩略图|`im.thumbnail((100,50))`|
|保存图片|`im.save("out.jpg")`|
|旋转图片（逆时针）|`im2 = im.rotate(90)`|
|裁剪区域|`crop_img = im.crop((10,10,100,80))`|

完整小示例：
```python
from PIL import Image
# 打开图片
im = Image.open("test.jpg")
print(im.size)
# 生成缩略图
im.thumbnail((300,200))
im.save("small.jpg")
```

---

## 2. ImageDraw（画笔工具，在图片上绘制内容）
**职责**：在已经存在的Image画布上画像素点、直线、矩形、圆形、文字
> 使用前提：必须绑定一个Image对象生成画笔
|常用功能|示例代码|
|---|---|
|创建画笔对象|`draw = ImageDraw.Draw(image)`|
|绘制单个像素点|`draw.point((50,50), fill=(255,0,0))`|
|画直线|`draw.line([(10,10),(100,100)], fill="red", width=2)`|
|画矩形框|`draw.rectangle([20,20,80,80], outline="blue")`|
|绘制文字|`draw.text((30,30),"Hello",font=font,fill=(0,0,0))`|

完整小示例：
```python
from PIL import Image, ImageDraw
img = Image.new("RGB",(150,100),"white")
draw = ImageDraw.Draw(img)
draw.point((20,20), fill=(255,0,0))
draw.rectangle([40,40,90,80], outline="green")
img.save("draw_demo.jpg")
```

---

## 3. ImageFont（字体工具，控制文字样式：字体文件、字号）
**职责**：专门给 `ImageDraw.text()` 提供字体与字号；不直接绘图，只是文字样式对象
|常用功能|示例代码|
|---|---|
|加载系统TrueType字体（自定义字号）|`font = ImageFont.truetype(r"C:\Windows\Fonts\arial.ttf",36)`|
|内置默认字体（免系统字体文件，兼容性最好）|`font = ImageFont.load_default(size=30)`|

完整小示例：
```python
from PIL import Image, ImageDraw, ImageFont
img = Image.new("RGB",(200,80),"white")
draw = ImageDraw.Draw(img)
# 加载字体
font = ImageFont.load_default(size=36)
draw.text((20,10),"Pillow Test", font=font, fill=(0,0,255))
img.save("font_demo.jpg")
```

---

## 4. ImageFilter（滤镜工具，给整张图片添加特效）
**职责**：对整张Image图片做批量像素处理，模糊、锐化、边缘检测等滤镜效果
|常用功能|示例代码|
|---|---|
|高斯模糊|`new_img = im.filter(ImageFilter.BLUR)`|
|锐化|`new_img = im.filter(ImageFilter.SHARPEN)`|
|轮廓描边（边缘提取）|`new_img = im.filter(ImageFilter.CONTOUR)`|
|浮雕效果|`new_img = im.filter(ImageFilter.EMBOSS)`|

完整小示例：
```python
from PIL import Image, ImageFilter
im = Image.open("test.jpg")
blur_img = im.filter(ImageFilter.BLUR)
blur_img.save("blur_demo.jpg")
```

# 整合对照总表（速查笔记版）
|模块|核心定位|能干什么|不能干什么|
|----|----|----|----|
|**Image**|图片画布本体|打开/新建/保存/缩放/裁剪旋转|**不能直接画线写字**|
|**ImageDraw**|绘图画笔|画点、线、矩形、文字|**不能创建画布，必须依附Image**|
|**ImageFont**|文字样式包|设置字体、字号|**不能独立写字，必须传给draw.text**|
|**ImageFilter**|图片滤镜特效|整图模糊、锐化、浮雕|**不能局部修改，整张图生效**|

# 完整串联示例（验证码迷你完整版）
```python
import random
from PIL import Image, ImageDraw, ImageFont, ImageFilter

# 1.Image 创建画布
img = Image.new("RGB",(240,60),"white")
# 2.ImageDraw 创建画笔
draw = ImageDraw.Draw(img)
# 3.ImageFont 设置字体
font = ImageFont.load_default(size=36)
draw.text((10,10),"ABCD",font=font,fill=(50,50,50))
# 4.ImageFilter 添加模糊滤镜
img = img.filter(ImageFilter.BLUR)
# Image保存
img.save("final.jpg")
```
